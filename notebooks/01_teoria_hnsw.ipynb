# T1 · Simulación reproducible de capas HNSW
Ejecución local real con Python; semilla 22, un millón de vectores.

```python
import random, math
rng = random.Random(22)
N, M = 1_000_000, 16
counts = [0, 0, 0]
for _ in range(N):
    level = math.floor(-math.log(1.0-rng.random()) / math.log(M))
    for k in (1, 2, 3):
        counts[k-1] += level >= k
for k, observed in enumerate(counts, 1):
    expected = N * M**(-k)
    print(f"capa >= {k}: esperado={expected:.6f}; simulado={observed}; diferencia={observed-expected:.6f}")
```
